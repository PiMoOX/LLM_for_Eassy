```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

client = OpenAI(
    api_key=os.getenv("ZAI_API_KEY"),
    base_url=os.getenv("ZAI_BASE_URL"),
    timeout=300,
    max_retries=3,
)

MODEL = "glm-4.5-flash"
```

# 📝 generate_draft Function
Objective: Write a function called generate_draft that takes in a string prompt and uses a language model to generate a complete draft essay.

Inputs:

prompt (str): The essay question or writing prompt.
model (str, optional): The model identifier to use. Defaults to "openai:gpt-5.1".
Output:

A string representing the full draft of the essay.

```python

def generate_draft(prompt: str, model: str = MODEL) -> str:
    instruction = f"""
Write a well-structured draft essay in Persian in response to the following prompt.
The draft should include an introduction, body, and conclusion.

Prompt:
{prompt}
"""
    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": instruction}],
        temperature=0.7,
    )
    return response.choices[0].message.content
```

```python
draft = generate_draft("تأثیر هوش مصنوعی بر آموزش", model="glm-4.5-flash")
print(draft)
```

### 🔍 reflect_on_draft Function
Objective: Write a function called reflect_on_draft that takes a previously generated essay draft and uses a language model to provide constructive feedback.

Inputs:

draft (str): The essay text to reflect on.
model (str, optional): The model identifier to use. Defaults to "openai:gpt-5.1".
Output:

A string with feedback in paragraph form.

```python
def reflect_on_draft(draft: str, model: str = MODEL) -> str:
    instruction = f"""
پیش‌نویس مقالهٔ زیر را نقادانه بررسی کن.
زمینه‌های قابل بهبود در ساختار، وضوح، قوت استدلال و سبک را مشخص کن.
بازخورد سازنده را به‌صورت پاراگراف بنویس.
هر کلمه یا جملهٔ غیرفارسی که در متن دیدی هم جداگانه ذکر کن.
تمام پاسخ تو باید فقط به زبان فارسی باشد و هیچ جملهٔ انگلیسی نداشته باشد.

پیش‌نویس:
{draft}
"""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": "تو یک ویراستار فارسی‌زبان هستی. همیشه فقط به فارسی پاسخ بده."},
            {"role": "user", "content": instruction},
        ],
        temperature=0.3,
    )
    return response.choices[0].message.content
```

### 🔁 revise_draft Function
Objective: Implement a function called revise_draft that improves a given essay draft based on feedback from a reflection step.

Inputs:

original_draft (str): The initial version of the essay.
reflection (str): Constructive feedback or critique on the draft.
model (str, optional): The model identifier to use. Defaults to "openai:gpt-5.1".
Output:

A string containing the revised and improved essay.

```python
def revise_draft(draft: str, critique: str, model: str = MODEL) -> str:
    instruction = f"""
مقالهٔ زیر را فقط با رفع نقدهای داده‌شده بازنویسی کن.
آنچه خوب است حفظ شود.
نسخهٔ نهایی باید کاملاً به فارسی باشد و هیچ کلمه یا نویسه‌ای از زبان‌ها و خط‌های دیگر نداشته باشد.
فقط متن نهایی مقاله را بنویس، بدون توضیح اضافه.

مقاله:
{draft}

نقدها:
{critique}
"""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": "تو یک نویسندهٔ حرفه‌ای فارسی هستی. همیشه فقط به فارسی بنویس."},
            {"role": "user", "content": instruction},
        ],
        temperature=0.7,
    )
    return response.choices[0].message.content
```

## 🧪 Test the Reflective Writing Workflow
Use the functions you implemented to simulate the complete writing workflow:

Generate a draft in response to the essay prompt.
Reflect on the draft to identify improvements.
Revise the draft using the feedback.

```python
MODEL = "glm-4.5-flash"


essay_prompt = "تأثیر هوش مصنوعی بر آموزش"

# Agente 1 – Draft
draft = generate_draft(essay_prompt, model=MODEL)
print("📝 Draft:\n")
print(draft)

# Agente 2 – Reflection
feedback = reflect_on_draft(draft, model=MODEL)
print("\n🧠 Feedback:\n")
print(feedback)

# Agente 3 – Revision
revised = revise_draft(draft, feedback, model=MODEL)
print("\n✍️ Revised:\n")
print(revised)
```

```python
def refine_with_user(essay: str, user_request: str, model: str = MODEL) -> str:
  
    instruction = f"""
Revise the essay below according to the user's request.
Apply only the changes the user asked for and keep the rest of the text as it is.
If the user asks for something to be added, place it in the most suitable part of the essay.
The final essay must be written entirely in Persian (Farsi script). Do not use words or
characters from any other language or script.
Output only the complete essay text, with no extra explanation.

Essay:
{essay}

User request:
{user_request}
"""
    response = client.chat.completions.create(
        model=model,
        messages=[
            # The system message fixes the role and the output language
            {"role": "system", "content": "You are a professional Persian writer. Always write only in Persian."},
            {"role": "user", "content": instruction},
        ],
        temperature=0.7,  # moderate creativity for rewriting
    )
    return response.choices[0].message.content
```

```python
def interactive_refine(essay: str, model: str = MODEL):
    """
    Interactive refinement loop.

    Commands:
        <empty Enter> / done / exit / q  -> finish
        undo                             -> go back one version
        show                             -> display the current version again
        anything else                    -> sent to the model as a change request

    Returns:
        (final_essay, versions), where versions is the full history.
    """
    END_WORDS = {"", "done", "exit", "q"}

    versions = [essay]  # versions[0] is the original; the last item is the current one

    def current():
        return versions[-1]

    def display(title, color):
        show_output(title, current(), background=color, text_color="#222222")

    print("You can now refine the essay with your own instructions.")
    print('Examples: "Add a section about education in Iran" or "Make the tone more formal"')
    print('Commands: undo = go back | show = redisplay | done (or empty Enter) = finish\n')

    while True:
        request = input("Your request: ").strip()
        command = request.lower()

        # Finish
        if command in END_WORDS:
            print("Finished.")
            break

        # Undo: remove the latest version (the original can never be removed)
        if command == "undo":
            if len(versions) > 1:
                versions.pop()
                print(f"Went back to version {len(versions) - 1}.")
                display(f"Version {len(versions) - 1} (restored)", "#f3e5f5")
            else:
                print(">>> Nothing to undo yet: send at least one change request first.")
            continue

        # Show the current version again
        if command == "show":
            display(f"Version {len(versions) - 1} (current)", "#fff8dc")
            continue

        # Normal request: ask the model to apply it to the current essay
        try:
            new_version = refine_with_user(current(), request, model=model)
        except Exception as e:
            # Network errors, rate limits, etc. must not crash the loop
            print(">>> Request failed:", e)
            continue

        versions.append(new_version)
        display(f"Version {len(versions) - 1} (after your request)", "#e8f5e9")

    return current(), versions
```

```python
final_essay, all_versions = interactive_refine(revised, model=MODEL)
```
